# Задачи по вероятности

## Задача 1. Монеты
Шанс орла 1/2, значит два орла подряд: 1/2 · 1/2 = **1/4**

## Задача 2. Монета и кубик
Орёл 1/2, шестёрка 1/6: 1/2 · 1/6 = **1/12**

## Задача 3. Два стрелка
а) 0,8 · 0,5 = **0,4**

б) Проще через противоположное событие «оба промахнулись»: 0,2 · 0,5 = 0,1. Тогда 1 − 0,1 = **0,9**

## Задача 7. Ровно одно попадание
Промахи: 0,5; 0,4; 0,2. Перебираем, кто один попал:
- 1-й: 0,5 · 0,4 · 0,2 = 0,04
- 2-й: 0,5 · 0,6 · 0,2 = 0,06
- 3-й: 0,5 · 0,4 · 0,8 = 0,16

Сумма: 0,04 + 0,06 + 0,16 = **0,26**

## Задача 8. Работа системы
Блок исправен: 0,9 · 0,9 = 0,81, сломан: 0,19.
Устройство не работает, только если сломаны оба блока: 0,19 · 0,19 = 0,0361.
Работает: 1 − 0,0361 = **0,9639**

## Задача 9. Известно, что кто-то попал
- Оба попали: 0,8 · 0,6 = 0,48
- Хотя бы один попал: 1 − 0,2 · 0,4 = 0,92

Среди случаев с «хотя бы одним попаданием» нужна доля случаев «оба попали»: 0,48 / 0,92 = **12/23 ≈ 0,52**

## Задача 4. Две коробки
Возьмём 8 равновероятных вариантов (2 коробки × 4 шара). Красный шар: 3 варианта в 1-й коробке и 1 во 2-й. Из 4 красных вариантов 3 относятся к 1-й коробке: **3/4**

## Задача 5. Необычная монета
Считаем стороны с орлом: у честной монеты 1, у необычной 2. Всего 3, из них 2 у необычной: **2/3**

## Задача 6. Спам-фильтр
Письмо помечено в двух случаях:
- оно спам и фильтр его пометил: 0,5 · 0,8 = 0,4
- оно обычное и фильтр ошибочно его пометил: 0,5 · 0,2 = 0,1

Вероятность, что письмо помечено: 0,4 + 0,1 = 0,5.
Нужна доля случаев «спам» среди помеченных: 0,4 / 0,5 = **0,8**

## Задача 10. Дискретное распределение
Математическое ожидание по определению: сумма «значение · вероятность».

E[X] = −2 · 0,2 + 0 · 0,3 + 3 · 0,4 + 5 · 0,1 = −0,4 + 0 + 1,2 + 0,5 = **1,3**

Для 2X + 1 значения становятся: 2·(−2)+1 = −3, 2·0+1 = 1, 2·3+1 = 7, 2·5+1 = 11 (вероятности те же).

E[2X + 1] = −3 · 0,2 + 1 · 0,3 + 7 · 0,4 + 11 · 0,1 = −0,6 + 0,3 + 2,8 + 1,1 = **3,6**
