# Задача A Второй тур

```python
import sys
import pandas as pd

items = pd.read_csv("items_A.csv")
edges = pd.read_csv("also_viewed_A.csv")
```

```python
count = 0
for i in range(len(items)):
    if items["category"][i] == "phones" and items["rating"][i] >= 4.5 and items["in_stock"][i] == 1:
        count += 1

print("A1:", count)
```

```python
sums = {}
counts = {}
for i in range(len(items)):
    if items["category"][i] == "laptops":
        brand = items["brand"][i]
        price = items["price"][i]
        if brand not in sums:
            sums[brand] = 0
            counts[brand] = 0
        sums[brand] += price
        counts[brand] += 1

best_brand = ""
best_avg = -1
for brand in sums:
    avg = sums[brand] / counts[brand]
    if avg > best_avg:
        best_avg = avg
        best_brand = brand

print("A2:", best_brand)
```

```python
count = 0
for i in range(len(items)):
    if items["in_stock"][i] == 1 and items["rating"][i] >= 4.5 and items["price"][i] >= 50000:
        count += 1

print("A3:", count)
```

```python
category_of = {}
for i in range(len(items)):
    category_of[items["item_id"][i]] = items["category"][i]

seen_in_to = set()
for x in edges["item_to"]:
    seen_in_to.add(x)

cnt = {}
for c in items["category"]:
    cnt[c] = 0
for item in seen_in_to:
    cnt[category_of[item]] += 1

categories = sorted(cnt.keys())
f = open("answer4.csv", "w")
f.write("category,cnt\n")
for c in categories:
    f.write(c + "," + str(cnt[c]) + "\n")
f.close()
print("A4:")
for c in categories:
    print(c, cnt[c])
```

```python
sys.setrecursionlimit(10000)

neighbors = {}
for item in items["item_id"]:
    neighbors[item] = []
for i in range(len(edges)):
    a = edges["item_from"][i]
    b = edges["item_to"][i]
    neighbors[a].append(b)
    neighbors[b].append(a)

visited = set()
island = []

def dfs(v):
    visited.add(v)
    island.append(v)
    for u in neighbors[v]:
        if u not in visited:
            dfs(u)

count = 0
for start in neighbors:
    if start in visited:
        continue
    island = []
    dfs(start)
    has_phones = False
    has_accessories = False
    for item in island:
        if category_of[item] == "phones":
            has_phones = True
        if category_of[item] == "accessories":
            has_accessories = True
    if has_phones and has_accessories:
        count += 1

print("A5:", count)
```

# Задача D Первый тур

```python
xs = [0, 0, 0, 0, 4, 4, 6, 6, 9, 9, 9, 9, 11, 11]
ys = [8, 7, 2, 0, 7, 4, 5, 1, 7, 4, 1, 0, 6, 2]
n = len(xs)
best = 10**9
for i in range(n):
    for j in range(i + 1, n):
        if xs[i] == xs[j]:
            continue
        a = (ys[j] - ys[i]) / (xs[j] - xs[i])
        b = ys[i] - a * xs[i]
        total = 0
        for k in range(n):
            err = abs(a * xs[k] + b - ys[k])
            total += err
        mae = total / n
        if mae < best:
            best = mae

print("%.6f" % best)
```

# Задача F Первый тур

```python
import random

rules = {
    "A": ["M", "D"],
    "E": ["T", "S", "M"],
    "I": ["T", "S", "M"],
    "T": ["L", "N"],
    "M": ["L", "N"],
    "N": ["A", "I"],
    "D": ["A", "I"],
    "S": ["E", "I", "D"],
    "L": ["E", "I", "D"],
}

random.seed(3256925993)
games = 10000000
total_steps = 0

for g in range(games):
    word = "DS"
    steps = 0
    while "ML" not in word:
        last = word[-1]
        new_letter = random.choice(rules[last])
        word = word + new_letter
        steps += 1
    total_steps += steps

print("%.2f" % (total_steps / games))
```
